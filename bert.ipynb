```python
import pandas as pd
import numpy as np

# Load data
df = pd.read_csv("Data/paragraphs.csv")

# Keep valid paragraphs
df = df.dropna(subset=["text"]).copy()
df["text"] = df["text"].astype(str).str.strip()
df = df[df["text"] != ""].reset_index(drop=True)

docs = df["text"].tolist()

print("Number of paragraphs:", len(docs))
print(docs[0])
```

```python
from sentence_transformers import SentenceTransformer

# Load Portuguese Sentence Transformer
embedding_model = SentenceTransformer("tcepi/sts_bertimbau")

# Generate paragraph embeddings
embeddings = embedding_model.encode(
    docs,
    show_progress_bar=True,
    batch_size=32,
    convert_to_numpy=True
)

print("Embeddings shape:", embeddings.shape)
```

```python
from bertopic import BERTopic
from umap import UMAP
from hdbscan import HDBSCAN
from sklearn.feature_extraction.text import CountVectorizer

# Dimensionality reduction
umap_model = UMAP(
    n_neighbors=15,
    n_components=5,
    min_dist=0.0,
    metric="cosine",
    random_state=42
)

# Clustering
hdbscan_model = HDBSCAN(
    min_cluster_size=20,
    min_samples=5,
    metric="euclidean",
    cluster_selection_method="eom",
    prediction_data=True
)

# Portuguese stopwords
vectorizer_model = CountVectorizer(
    stop_words="portuguese",
    ngram_range=(1, 2),
    min_df=5
)

# BERTopic
topic_model = BERTopic(
    embedding_model=embedding_model,
    umap_model=umap_model,
    hdbscan_model=hdbscan_model,
    vectorizer_model=vectorizer_model,
    language="portuguese",
    calculate_probabilities=True,
    verbose=True
)

# Fit model
topics, probabilities = topic_model.fit_transform(
    docs,
    embeddings
)
```

```python
# Topic overview
topic_info = topic_model.get_topic_info()

print(topic_info)
```
